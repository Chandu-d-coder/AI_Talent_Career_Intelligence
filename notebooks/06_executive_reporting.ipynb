```python
import warnings
import platform
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configure execution environment and suppress warnings
warnings.filterwarnings("ignore")

# Optimize pandas display configurations for comprehensive data inspection.
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", None)

# Log framework metadata and verification metrics.

print("Environment Information")
print("-" * 22)
print(f"Python Version : {platform.python_version()}")
print(f"Pandas Version : {pd.__version__}")
print(f"NumPy Version  : {np.__version__}")
```

```python
# Load target dataset

DATA_PATH = "../../data/processed/cleaned_ai_jobs.csv"
df = pd.read_csv(DATA_PATH)
print("\nDataset Loaded Successfully")

# Execute descriptive matrix shape verification.

print("\nDataset Shape")
print("-" * 16)

total_rows, total_columns = df.shape

print(f"Rows    : {total_rows:,}")
print(f"Columns : {total_columns}")
print("-" * 16)

# Display a quick preview of the dataset schema.

display(df.head())
```

```python
# Executive KPI Dashboard

executive_kpis = pd.DataFrame({
    "Metric": [
        "Total Records","Countries","Industries","AI Specializations",
        "Average Salary (USD)","Maximum Salary (USD)","Average Bonus (USD)",
        "Average Career Growth","Average Job Security",
        "Average Employee Satisfaction","Average Offer Acceptance",
        "Average Job Openings"
    ],
    "Value": [
        len(df),
        df["country"].nunique(),
        df["industry"].nunique(),
        df["ai_specialization"].nunique(),
        round(df["salary_usd"].mean(),2),
        round(df["salary_usd"].max(),2),
        round(df["bonus_usd"].mean(),2),
        round(df["career_growth_score"].mean(),2),
        round(df["job_security_score"].mean(),2),
        round(df["employee_satisfaction"].mean(),2),
        round(df["offer_acceptance_rate"].mean(),2),
        round(df["job_openings"].mean(),2)
    ]
})

executive_kpis
```

## Chart 1: Executive KPI Dashboard

```python
kpi_values = {
    "Salary": df["salary_usd"].mean(),
    "Career Growth": df["career_growth_score"].mean(),
    "Job Security": df["job_security_score"].mean(),
    "Employee Satisfaction": df["employee_satisfaction"].mean(),
    "Offer Acceptance": df["offer_acceptance_rate"].mean(),
    "Skill Demand": df["skill_demand_score"].mean()
}

fig, ax = plt.subplots(figsize=(12,6))

sns.barplot(
    x=list(kpi_values.keys()),
    y=list(kpi_values.values()),
    color="steelblue",
    ax=ax
)

ax.set_title("Executive Workforce KPI Dashboard", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("")
ax.set_ylabel("Average Score")
ax.grid(axis="y", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

## Chart 2: Global Workforce Distribution

```python
fig, ax = plt.subplots(figsize=(12,6))

sns.countplot(
    data=df,
    x="experience_level",
    order=["Entry","Mid","Senior","Lead"],
    color="seagreen",
    ax=ax
)

ax.set_title("Global AI Workforce Distribution by Experience Level", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Experience Level")
ax.set_ylabel("Number of Professionals")
ax.grid(axis="y", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

## Chart 3: AI Workforce by Country

```python
country_counts = df["country"].value_counts().reset_index()
country_counts.columns = ["country","count"]

fig, ax = plt.subplots(figsize=(12,6))

sns.barplot(
    data=country_counts,
    x="count",
    y="country",
    color="mediumpurple",
    ax=ax
)

ax.set_title("AI Workforce Distribution by Country", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Number of Professionals")
ax.set_ylabel("Country")
ax.grid(axis="x", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

# Salary Intelligence Summary

## Executive Overview

Salary Intelligence analysis examined global compensation patterns across countries, industries, AI specializations, education levels, and experience levels.

The analysis confirms that professional experience remains the strongest determinant of salary growth, while country and AI specialization also contribute significantly to compensation differences.

Organizations seeking to attract and retain AI talent should prioritize competitive compensation structures alongside clear career progression pathways.


## Executive Highlights

### Major Findings

- Lead professionals receive the highest average salaries.
- Salary consistently increases with years of experience.
- The USA offers the highest average AI compensation.
- Emerging AI domains such as LLM and Generative AI remain among the highest-paying specializations.
- Education level has relatively limited influence compared to professional experience.
- Bonus compensation closely follows salary growth.

### Business Impact

Organizations should maintain market-aligned salary structures to improve talent acquisition and retention while investing in high-demand AI specializations.


```python
salary_country = (
    df.groupby("country")["salary_usd"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=salary_country,
    x="salary_usd",
    y="country",
    color="steelblue",
    ax=ax
)

ax.set_title("Executive Salary Overview by Country", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Average Salary (USD)")
ax.set_ylabel("Country")
ax.grid(axis="x", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

# Career Growth Intelligence Summary

## Executive Overview

Career Growth Intelligence evaluated promotion speed, career growth scores, salary percentile, and workforce progression.

The findings indicate that promotion speed differentiates career progression more effectively than career growth scores, while AI specialization and experience significantly influence long-term advancement.

Organizations should build structured promotion frameworks supported by continuous learning and technical development.


## Executive Highlights

### Major Findings

- Promotion Speed increases consistently with experience.
- Career Growth Scores remain relatively stable across workforce segments.
- LLM and Generative AI specialists demonstrate stronger long-term career opportunities.
- Professional specialization influences promotion opportunities more than geographic location.
- Engineered career metrics successfully identify high-potential professionals.

### Business Impact

Career development strategies should emphasize skill progression, promotion transparency, and technical specialization rather than tenure alone.


```python
promotion_summary = (
    df.groupby("experience_level")["promotion_speed"]
    .mean()
    .reindex(["Entry", "Mid", "Senior", "Lead"])
    .reset_index()
)

fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=promotion_summary,
    x="experience_level",
    y="promotion_speed",
    color="seagreen",
    ax=ax
)

ax.set_title("Executive Promotion Speed Overview", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Experience Level")
ax.set_ylabel("Promotion Speed")
ax.grid(axis="y", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

# Workforce Risk Intelligence Summary

## Executive Overview

Workforce Risk Intelligence analyzed employment stability using Job Security, Layoff Risk, Automation Risk, and AI Adoption.

The results indicate that workforce stability depends primarily on organizational resilience and employee experience rather than automation alone.

Senior professionals consistently demonstrate stronger employment stability across industries.


## Executive Highlights

### Major Findings

- Job Security increases significantly with experience.
- Layoff Risk is the strongest predictor of workforce stability.
- AI Adoption has limited direct impact on Job Security.
- Employee Satisfaction positively contributes to workforce resilience.
- Technology-driven organizations maintain strong workforce stability.

### Business Impact

Organizations should invest in employee retention, internal career development, and workforce planning to strengthen long-term organizational resilience.


```python
job_security_summary = (
    df.groupby("experience_level")["job_security_score"]
    .mean()
    .reindex(["Entry", "Mid", "Senior", "Lead"])
    .reset_index()
)

fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=job_security_summary,
    x="experience_level",
    y="job_security_score",
    color="indianred",
    ax=ax
)

ax.set_title("Executive Job Security Overview", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Experience Level")
ax.set_ylabel("Job Security Score")
ax.grid(axis="y", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

# Talent Market Intelligence Summary

## Executive Overview

Talent Market Intelligence examined global hiring demand, skill demand, job openings, recruitment competitiveness, and market opportunities across the AI industry.

The analysis indicates that AI talent demand remains consistently strong across countries and industries, while emerging AI specializations continue to drive recruitment demand. Organizations increasingly compete for professionals possessing advanced AI capabilities, particularly in high-growth technical domains.

The engineered **Talent Demand Index** provides a holistic view of market attractiveness by combining hiring demand, AI adoption, and workforce opportunity indicators.


## Executive Highlights

### Major Findings

- AI hiring demand remains consistently high across global markets.
- LLM, NLP, Generative AI, and Reinforcement Learning continue to attract strong employer demand.
- AI Software Engineers and Machine Learning Engineers experience the highest recruitment activity.
- Skill Demand is a stronger indicator of market opportunity than geographic location.
- Healthcare, Technology, and Finance remain among the largest AI hiring industries.

### Business Impact

Organizations should continuously invest in workforce development and proactive talent acquisition strategies to remain competitive in rapidly evolving AI markets.


```python
talent_summary=(
    df.groupby("ai_specialization")["skill_demand_score"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

fig, ax = plt.subplots(figsize=(12,6))

sns.barplot(data=talent_summary,
            x="skill_demand_score",
            y="ai_specialization",
            color="steelblue",
            ax=ax)

ax.set_title("Executive AI Skill Demand Overview", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Average Skill Demand Score")
ax.set_ylabel("AI Specialization")
ax.grid(axis="x", linestyle="--", alpha=0.3)
sns.despine()
plt.tight_layout()
plt.show()
```

# Employee Experience Intelligence Summary

## Executive Overview

Employee Experience Intelligence evaluated workforce wellbeing using employee satisfaction, work-life balance, working hours, vacation policies, and company ratings.

The analysis demonstrates that employee satisfaction is primarily driven by compensation, job security, and healthy work-life balance. Flexible work arrangements alone do not significantly improve employee experience without supportive organizational practices.

The Employee Experience Index developed in this project enables organizations to monitor workforce wellbeing using multiple performance dimensions rather than isolated satisfaction metrics.


## Executive Highlights

### Major Findings

- Employee Satisfaction remains consistently high across the AI workforce.
- Salary and Job Security are the strongest contributors to employee satisfaction.
- Longer working hours reduce employee wellbeing.
- Work-Life Balance positively influences long-term employee engagement.
- Company Rating alone has limited influence on employee satisfaction.

### Business Impact

Organizations should improve employee experience by balancing competitive compensation, career stability, workload management, and employee wellbeing initiatives.


```python
employee_summary=(
    df.groupby("work_mode")["employee_satisfaction"]
    .mean()
    .reset_index()
)

fig, ax = plt.subplots(figsize=(10,6))

sns.barplot(data=employee_summary,
            x="work_mode",
            y="employee_satisfaction",
            color="seagreen",
            ax=ax)

ax.set_title("Executive Employee Satisfaction Overview", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Work Mode")
ax.set_ylabel("Employee Satisfaction")
ax.grid(axis="y", linestyle="--", alpha=0.3)
sns.despine()
plt.tight_layout()
plt.show()
```

# Recruitment Intelligence Summary

## Executive Overview

Recruitment Intelligence examined offer acceptance, recruitment efficiency, hiring difficulty, and candidate attraction across the AI workforce.

The analysis shows that AI recruitment remains healthy across industries, with consistently high offer acceptance rates. Recruitment outcomes depend on multiple organizational factors rather than compensation alone, emphasizing the importance of employer branding and candidate experience.

The engineered Recruitment Success Index and Candidate Attraction Score provide organizations with practical measures for evaluating hiring performance.


## Executive Highlights

### Major Findings

- Offer Acceptance remains consistently high across countries and industries.
- Salary alone does not determine candidate acceptance.
- Employer attractiveness contributes more to recruitment success than compensation alone.
- AI recruitment remains competitive despite increasing market demand.
- Efficient recruitment processes improve hiring effectiveness.

### Business Impact

Organizations should strengthen employer branding, simplify recruitment workflows, and enhance candidate experience to improve long-term hiring success.


```python
offer_summary=(
    df.groupby("country")["offer_acceptance_rate"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

fig, ax = plt.subplots(figsize=(12,6))

sns.barplot(data=offer_summary,
            x="offer_acceptance_rate",
            y="country",
            color="mediumpurple",
            ax=ax)

ax.set_title("Executive Offer Acceptance Overview", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Offer Acceptance Rate (%)")
ax.set_ylabel("Country")
ax.grid(axis="x", linestyle="--", alpha=0.3)
sns.despine()
plt.tight_layout()
plt.show()
```
