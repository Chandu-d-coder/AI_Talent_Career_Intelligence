```python
import platform
import numpy as np
import pandas as pd
import matplotlib
from matplotlib.ticker import StrMethodFormatter
import matplotlib.pyplot as plt
import seaborn as sns

# Optimize pandas display configurations for comprehensive data inspection.
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", None)

# Log framework metadata and verification metrics.
print("Environment Information")
print("-" * 22)
print(f"Python Version     : {platform.python_version()}")
print(f"Pandas Version     : {pd.__version__}")
print(f"NumPy Version      : {np.__version__}")
print(f"Matplotlib Version : {matplotlib.__version__}")
print(f"Seaborn Version    : {sns.__version__}")
```

```python
# Load target dataset

DATA_PATH = "../../data/processed/cleaned_ai_jobs.csv"
df = pd.read_csv(DATA_PATH)
print("\nDataset Loaded Successfully")

# Execute descriptive matrix shape verification.

print("\nDataset Shape")
print("-" * 16)

total_rows, total_columns = df.shape

print(f"Rows    : {total_rows:,}")
print(f"Columns : {total_columns}")
print("-" * 16)

# Display a quick preview of the dataset schema.

display(df.head())
```

```python
# Create the figure and axes.
fig, ax = plt.subplots(figsize=(15, 7), dpi=150)

# Apply a clean visualization theme.
sns.set_theme(
    style="whitegrid",
    context="notebook",
    font="DejaVu Sans"
)

# Plot the salary distribution.
sns.histplot(
    data=df,
    x="salary_usd",
    bins="fd",
    kde=True,
    stat="count",
    color="#4C72B0",
    edgecolor="white",
    linewidth=0.8,
    alpha=0.90,
    line_kws={
        "linewidth": 2.5,
        "color": "#1F4E79"
    },
    ax=ax
)

# Add a reference line for the median salary.
median_salary = df["salary_usd"].median()

ax.axvline(
    median_salary,
    color="#C44E52",
    linestyle="--",
    linewidth=2
)

# Configure the chart title and axis labels.
ax.set_title(
    "Distribution of Annual Salaries Across AI Job Records",
    fontsize=18,
    fontweight="bold",
    pad=20
)

ax.set_xlabel(
    "Annual Salary (USD)",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

ax.set_ylabel(
    "Number of Job Records",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

# Format the x-axis as currency.
ax.xaxis.set_major_formatter(StrMethodFormatter("${x:,.0f}"))

# Format tick labels.
ax.tick_params(
    axis="both",
    labelsize=11
)

# Improve grid styling.
ax.grid(
    axis="y",
    linestyle="--",
    linewidth=0.8,
    alpha=0.35
)

ax.grid(
    axis="x",
    visible=False
)

# Remove unnecessary spines.
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Optimize layout.
fig.tight_layout()

# Display the visualization.
plt.show()
```

```python
# Create Engineered Workforce Risk Features

df["workforce_stability_index"] = (
    (
        df["job_security_score"] +
        df["ai_adoption_score"] +
        (100 - df["layoff_risk"] * 100)
    ) / 3
).round(2)

df["automation_resilience_score"] = (
    (
        df["job_security_score"] +
        (100 - df["automation_risk"])
    ) / 2
).round(2)

df["risk_exposure_index"] = (
    (
        (df["layoff_risk"] * 100) +
        df["automation_risk"]
    ) / 2
).round(2)

df["ai_workforce_readiness_score"] = (
    (
        df["job_security_score"] +
        df["ai_adoption_score"] +
        df["skill_demand_score"]
    ) / 3
).round(2)

df["stability_risk_ratio"] = (
    df["job_security_score"] /
    (df["layoff_risk"] * 100 + 1)
).round(2)
```

```python
if "recruitment_success_index" not in df.columns:
    df["recruitment_success_index"]=(
        (
            df["offer_acceptance_rate"]+
            (100-df["hiring_difficulty_score"])+
            (100-(df["interview_rounds"]*10))
        )/3
    ).round(2)

if "candidate_attraction_score" not in df.columns:
    df["candidate_attraction_score"]=(
        (
            df["offer_acceptance_rate"]+
            df["salary_percentile"]+
            df["company_rating"]*20
        )/3
    ).round(2)

recruitment_corr=(
    df[[
        "offer_acceptance_rate",
        "salary_usd",
        "bonus_usd",
        "interview_rounds",
        "hiring_difficulty_score",
        "career_growth_score",
        "company_rating",
        "employee_satisfaction",
        "recruitment_success_index",
        "candidate_attraction_score"
    ]].corr()
)

fig, ax = plt.subplots(figsize=(10,8))

sns.heatmap(
    recruitment_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink":0.8},
    ax=ax
)

ax.set_title("Recruitment Intelligence Correlation Matrix", fontsize=16, fontweight="bold", pad=15)

plt.tight_layout()
plt.show()
```
