### Analysis 1 - Salary by Experience Level

This analysis evaluates how salary changes as professionals progress through different experience levels.

```python
# Calculate descriptive salary statistics across experience levels

salary_by_experience_level = (
    df.groupby("experience_level")
      .agg(
          job_posting_count=("salary_usd", "count"),
          average_salary=("salary_usd", "mean"),
          median_salary=("salary_usd", "median"),
          minimum_salary=("salary_usd", "min"),
          maximum_salary=("salary_usd", "max"),
          salary_standard_deviation=("salary_usd", "std")
      )
      .round(2)
      .sort_values(by="average_salary", ascending=False)
)

salary_by_experience_level
```

## Workforce Salary Progression

## Workforce Ranking

Dense ranking assigns the same rank to equal values without leaving gaps in the ranking sequence.

This provides an alternative perspective on workforce performance.

```python
# Rank job roles based on average salary

average_salary_by_job_role = (
    df.groupby("job_role")["salary_usd"]
      .mean()
)

job_role_salary_rank = (
    average_salary_by_job_role
    .rank(
        ascending=False,
        method="dense"
    )
    .astype(int)
    .sort_values()
)

job_role_salary_rank
```

```python
# Rank countries based on average salary

country_salary_rank = (
    df.groupby("country")["salary_usd"]
      .mean()
      .sort_values(ascending=False)
      .rank(
          ascending=False,
          method="dense"
      )
      .head(5)
)

country_salary_rank
```

## Salary Distribution Analysis

```python
# Calculate salary percentiles

salary_percentile_values = np.percentile(
    df["salary_usd"],
    [10, 30, 50, 75, 100]
)

salary_percentile_summary = pd.DataFrame({
    "Percentile": [
        "10%",
        "30%",
        "50%",
        "75%",
        "100%"
    ],
    "Salary": salary_percentile_values
})

salary_percentile_summary
```

```python
# Calculate bonus percentiles

bonus_percentile_values = np.percentile(
    df["bonus_usd"],
    [10, 30, 50, 75, 100]
)

bonus_percentile_summary = pd.DataFrame({
    "Percentile": [
        "10%",
        "30%",
        "50%",
        "75%",
        "100%"
    ],
    "Bonus": bonus_percentile_values,
    "Country": country_salary_rank
})

bonus_percentile_summary
```

```python
# Create salary quartile categories

df["salary_quartile"] = pd.qcut(
    df["salary_usd"],
    q=4,
    labels=[
        "Low",
        "Mid",
        "High",
        "Best"
    ]
)

df["salary_percentile"].value_counts(
    normalize=True
).mul(100)
```

## Compensation Spread

Spread measures the distance between the highest and lowest values within a distribution.

These metrics provide a quick indication of compensation inequality across the workforce.

```python
# Calculate salary and bonus spread

salary_spread = (
    df["salary_usd"].max() - df["salary_usd"].min()
)

bonus_spread = (
    df["bonus_usd"].max() - df["bonus_usd"].min()
)

distribution_spread_summary = pd.DataFrame({
    "Metric": [
        "Salary Spread",
        "Bonus Spread"
    ],
    "Value": [
        salary_spread,
        bonus_spread
    ]
})

distribution_spread_summary
```

## Statistical Variability

```python
# Calculate salary variance

salary_variance = np.var(
    df["salary_usd"]
).round(3)

salary_variance
```

```python
# Calculate salary standard deviation

salary_standard_deviation = np.std(
    df["salary_usd"]
).round(3)

salary_standard_deviation
```

Create new compensation metrics to better understand bonus contribution, compensation growth, and salary efficiency.

```python
# Calculate salary earned per year of professional experience

df["salary_per_experience_year"] = np.where(
    df["experience_years"] == 0,
    df["salary_usd"],
    (df["salary_usd"] / df["experience_years"]).round(2)
)

# Review the distribution of salary per experience year

salary_per_experience_year_summary = (
    df["salary_per_experience_year"]
    .describe()
    .round(2)
)

salary_per_experience_year_summary
```

Develop workforce metrics that measure salary value relative to market demand and overall compensation ranking.

# Talent Market Intelligence - Feature Engineering

This notebook creates engineered talent market metrics to measure hiring demand, market opportunities, recruitment competitiveness, and skill value. Each engineered feature is introduced with a business question followed by its implementation.


## Talent Demand Index

**Business Question:** Evaluate **talent demand index** to derive actionable talent market intelligence, identify high-opportunity segments, and support strategic recruitment planning.

```python
# Talent Demand Index

df["talent_demand_index"] = (
    (
        df["job_openings"] +
        df["skill_demand_score"] +
        df["ai_adoption_score"]
    ) / 3
).round(2)
```

## Hiring Efficiency Score

**Business Question:** Evaluate **hiring efficiency score** to derive actionable talent market intelligence, identify high-opportunity segments, and support strategic recruitment planning.

```python
# Hiring Efficiency Score

df["hiring_efficiency_score"] = (
    df["job_openings"] /
    (df["interview_rounds"] + 1)
).round(2)
```

## Market Opportunity Index

**Business Question:** Evaluate **market opportunity index** to derive actionable talent market intelligence, identify high-opportunity segments, and support strategic recruitment planning.

```python
# Market Opportunity Index

df["market_opportunity_index"] = (
    (
        df["job_openings"] +
        df["career_growth_score"] +
        df["salary_percentile"]
    ) / 3
).round(2)
```

## Skill Premium Index

**Business Question:** Evaluate **skill premium index** to derive actionable talent market intelligence, identify high-opportunity segments, and support strategic recruitment planning.

```python
# Skill Premium Index

df["skill_premium_index"] = (
    (
        df["skill_demand_score"] *
        df["salary_percentile"]
    ) / 100
).round(2)
```

## Recruitment Competitiveness Score

**Business Question:** Evaluate **recruitment competitiveness score** to derive actionable talent market intelligence, identify high-opportunity segments, and support strategic recruitment planning.

```python
# Recruitment Competitiveness Score

df["recruitment_competitiveness_score"] = (
    (
        df["skill_demand_score"] +
        df["hiring_difficulty_score"] +
        (df["interview_rounds"] * 10)
    ) / 3
).round(2)
```
