# Employee Experience Intelligence

This notebook explores employee satisfaction, work-life balance, weekly working hours, vacation benefits, and company ratings to understand employee experience across the AI workforce.

## Employee Experience Overview

**Business Question:** What is the overall distribution of employee experience indicators across the AI workforce?

```python
# Employee Experience Overview

df[[
    "employee_satisfaction",
    "work_life_balance_score",
    "weekly_hours",
    "vacation_days",
    "company_rating"
]].describe().round(2)
```

## Employee Satisfaction by Work Mode

**Business Question:** How does employee satisfaction vary across work modes?

```python
# Employee Satisfaction by Work Mode

df.groupby("work_mode")[
    "employee_satisfaction"
].agg(["mean", "median", "min", "max"]).round(2)
```

## Work-Life Balance by Work Mode

**Business Question:** Which work mode provides the best work-life balance?

```python
# Work-Life Balance by Work Mode

df.groupby("work_mode")[
    "work_life_balance_score"
].mean().round(2)
```

## Employee Satisfaction by Country

**Business Question:** Which countries report the highest employee satisfaction?

```python
# Employee Satisfaction by Country

df.groupby("country")[
    "employee_satisfaction"
].mean().sort_values(ascending=False).round(2)
```

## Employee Satisfaction by Industry

**Business Question:** Which industries provide the best employee experience?

```python
# Employee Satisfaction by Industry

df.groupby("industry")[
    "employee_satisfaction"
].mean().sort_values(ascending=False).round(2)
```

## Employee Satisfaction by Company Rating

**Business Question:** Does company rating influence employee satisfaction?

```python
# Employee Satisfaction by Company Rating

df.groupby("company_rating")[
    "employee_satisfaction"
].mean().round(2)
```

# Employee Experience Intelligence - Relationship Analytics

## Correlation with Employee Satisfaction

**Business Question:** Which variables have the strongest relationship with employee satisfaction?

```python
# Correlation with Employee Satisfaction

employee_corr = (
    df[[
        "employee_satisfaction",
        "work_life_balance_score",
        "weekly_hours",
        "vacation_days",
        "company_rating",
        "career_growth_score",
        "salary_usd",
        "job_security_score"
    ]]
    .corr()["employee_satisfaction"]
    .sort_values(ascending=False)
    .round(2)
)

employee_corr
```

## Work Mode vs Work-Life Balance

**Business Question:** How does work-life balance vary across work modes and experience levels?

```python
# Work Mode vs Work-Life Balance

pd.pivot_table(
    df,
    values="work_life_balance_score",
    index="work_mode",
    columns="experience_level",
    aggfunc="mean"
).round(2)
```

## Company Rating vs Satisfaction

**Business Question:** How does company rating influence satisfaction?

```python
# Company Rating vs Satisfaction

pd.pivot_table(
    df,
    values="employee_satisfaction",
    index="company_rating",
    columns="work_mode",
    aggfunc="mean"
).round(2)
```

## Weekly Hours vs Employee Satisfaction

**Business Question:** How do weekly working hours affect satisfaction?

```python
# Weekly Hours vs Employee Satisfaction

pd.crosstab(
    index=df["weekly_hours"],
    columns=df["work_mode"],
    values=df["employee_satisfaction"],
    aggfunc="mean"
).round(2)
```

# Employee Experience Intelligence - Feature Engineering

## Employee Experience Index

**Business Question:** Can overall employee experience be summarized into a single index?

```python
# Employee Experience Index

df["employee_experience_index"] = (
    (
        df["employee_satisfaction"] +
        df["work_life_balance_score"] +
        df["company_rating"] * 20
    ) / 3
).round(2)
```

## Wellbeing Score

**Business Question:** How can employee wellbeing be measured?

```python
# Wellbeing Score

df["wellbeing_score"] = (
    (
        df["employee_satisfaction"] +
        df["work_life_balance_score"] +
        (100 - df["weekly_hours"])
    ) / 3
).round(2)
```

## Engineered Metrics Summary

**Business Question:** What is the statistical summary of the engineered metrics?

```python
# Engineered Metrics Summary

df[[
    "employee_experience_index",
    "wellbeing_score"
]].describe().round(2)
```

## Top Employee Experience

**Business Question:** Which professionals have the highest employee experience scores?

```python
# Top Employee Experience

df.nlargest(
    10,
    "employee_experience_index"
)[[
    "job_role",
    "country",
    "employee_experience_index"
]]
```
