# Recruitment Intelligence

This notebook contains the foundation analysis, relationship analytics, and feature engineering for Recruitment Intelligence.

## Recruitment Overview

**Business Question:** What are the overall recruitment metrics?

```python
df[[
    "offer_acceptance_rate",
    "interview_rounds",
    "hiring_difficulty_score"
]].describe().round(2)
```

## Offer Acceptance by Country

**Business Question:** How does offer acceptance vary across countries?

```python
df.groupby("country")["offer_acceptance_rate"].agg(["mean","median","min","max"]).round(2)
```

## Offer Acceptance by Job Role

**Business Question:** Which job roles have the highest offer acceptance?

```python
df.groupby("job_role")["offer_acceptance_rate"].mean().sort_values(ascending=False).round(2)
```

## Offer Acceptance by AI Specialization

**Business Question:** Which AI specializations have the highest offer acceptance?

```python
df.groupby("ai_specialization")["offer_acceptance_rate"].mean().sort_values(ascending=False).round(2)
```

## Offer Acceptance by Industry

**Business Question:** How does offer acceptance vary across industries?

```python
df.groupby("industry")["offer_acceptance_rate"].mean().sort_values(ascending=False).round(2)
```

# Recruitment Intelligence - Relationship Analytics

## Correlation with Offer Acceptance

**Business Question:** Which variables influence offer acceptance?

```python
recruitment_corr=(df[[
"offer_acceptance_rate","salary_usd","bonus_usd","interview_rounds",
"hiring_difficulty_score","career_growth_score","company_rating",
"employee_satisfaction"]].corr()["offer_acceptance_rate"].sort_values(ascending=False).round(2))
recruitment_corr
```

## Offer Acceptance by Experience Level

**Business Question:** How does offer acceptance vary by experience and work mode?

```python
pd.pivot_table(df,values="offer_acceptance_rate",index="experience_level",columns="work_mode",aggfunc="mean").round(2)
```

## Offer Acceptance by Industry

**Business Question:** How does offer acceptance vary by industry and experience?

```python
pd.pivot_table(df,values="offer_acceptance_rate",index="industry",columns="experience_level",aggfunc="mean").round(2)
```

## Hiring Difficulty by Job Role

**Business Question:** How does hiring difficulty vary across job roles?

```python
pd.crosstab(index=df["job_role"],columns=df["experience_level"],values=df["hiring_difficulty_score"],aggfunc="mean").round(2)
```

# Recruitment Intelligence - Feature Engineering

## Recruitment Success Index

**Business Question:** Can recruitment effectiveness be summarized into one metric?

```python
df["recruitment_success_index"]=((df["offer_acceptance_rate"]+(100-df["hiring_difficulty_score"])+(100-(df["interview_rounds"]*10)))/3).round(2)
```

## Candidate Attraction Score

**Business Question:** Can candidate attraction be summarized into one metric?

```python
df["candidate_attraction_score"]=((df["offer_acceptance_rate"]+df["salary_percentile"]+df["company_rating"]*20)/3).round(2)
```

## Recruitment Metrics Summary

**Business Question:** What is the summary of engineered metrics?

```python
df[["recruitment_success_index","candidate_attraction_score"]].describe().round(2)
```

## Top Recruitment Success

**Business Question:** Which records have the highest recruitment success index?

```python
df.nlargest(10,"recruitment_success_index")[["job_role","country","recruitment_success_index"]]
```
