# Career Growth Intelligence - Feature Engineering

This notebook section creates engineered workforce metrics to quantify career progression, promotion efficiency, growth potential, and career momentum. Each engineered feature and its analysis is organized into separate markdown and code cells.

## Career Acceleration Index

**Business Question:** Evaluate **career acceleration index** to derive actionable workforce intelligence from engineered career growth metrics and identify high-performing talent segments.

```python
# Career Acceleration Index

df["career_acceleration_index"] = (
    df["career_growth_score"] /
    (df["experience_years"] + 1)
).round(2)
```

## Promotion Efficiency Index

**Business Question:** Evaluate **promotion efficiency index** to derive actionable workforce intelligence from engineered career growth metrics and identify high-performing talent segments.

```python
# Promotion Efficiency Index

df["promotion_efficiency"] = (
    df["promotion_speed"] /
    (df["experience_years"] + 1)
).round(2)
```

## Career Momentum Score

**Business Question:** Evaluate **career momentum score** to derive actionable workforce intelligence from engineered career growth metrics and identify high-performing talent segments.

```python
# Career Momentum Score

df["career_momentum_score"] = (
    (
        df["career_growth_score"] +
        df["promotion_speed"] +
        df["salary_percentile"]
    ) / 3
).round(2)
```

## Growth Potential Index

**Business Question:** Evaluate **growth potential index** to derive actionable workforce intelligence from engineered career growth metrics and identify high-performing talent segments.

```python
# Growth Potential Index

df["growth_potential_index"] = (
    (
        df["career_growth_score"] *
        df["skill_demand_score"]
    ) / 100
).round(2)
```

## Experience Efficiency Score

**Business Question:** Evaluate **experience efficiency score** to derive actionable workforce intelligence from engineered career growth metrics and identify high-performing talent segments.

```python
# Experience Efficiency Score

df["experience_efficiency_score"] = (
    df["career_growth_score"] /
    (df["experience_years"] + 1)
).round(2)
```

## Engineered Features Summary

**Business Question:** Evaluate **engineered features summary** to derive actionable workforce intelligence from engineered career growth metrics and identify high-performing talent segments.

```python
# Engineered Features Summary

df[[
    "career_acceleration_index",
    "promotion_efficiency",
    "career_momentum_score",
    "growth_potential_index",
    "experience_efficiency_score"
]].describe().round(2)
```
