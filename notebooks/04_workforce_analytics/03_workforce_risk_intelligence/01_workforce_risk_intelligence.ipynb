# Workforce Risk Intelligence - Feature Engineering

This notebook creates engineered workforce risk metrics to quantify organizational stability, automation resilience, workforce readiness, and overall risk exposure. Each engineered feature and its analysis is presented with a dedicated markdown and code cell.

## Workforce Stability Index

**Business Question:** Evaluate **workforce stability index** to derive advanced workforce risk intelligence, identify resilient talent segments, and support strategic workforce planning.

```python
# Workforce Stability Index

df["workforce_stability_index"] = (
    (
        df["job_security_score"] +
        df["ai_adoption_score"] +
        (100 - df["layoff_risk"] * 100)
    ) / 3
).round(2)
```

## Automation Resilience Score

**Business Question:** Evaluate **automation resilience score** to derive advanced workforce risk intelligence, identify resilient talent segments, and support strategic workforce planning.

```python
# Automation Resilience Score

df["automation_resilience_score"] = (
    (
        df["job_security_score"] +
        (100 - df["automation_risk"])
    ) / 2
).round(2)
```

## Risk Exposure Index

**Business Question:** Evaluate **risk exposure index** to derive advanced workforce risk intelligence, identify resilient talent segments, and support strategic workforce planning.

```python
# Risk Exposure Index

df["risk_exposure_index"] = (
    (
        (df["layoff_risk"] * 100) +
        df["automation_risk"]
    ) / 2
).round(2)
```

## AI Workforce Readiness Score

**Business Question:** Evaluate **ai workforce readiness score** to derive advanced workforce risk intelligence, identify resilient talent segments, and support strategic workforce planning.

```python
# AI Workforce Readiness Score

df["ai_workforce_readiness_score"] = (
    (
        df["job_security_score"] +
        df["ai_adoption_score"] +
        df["skill_demand_score"]
    ) / 3
).round(2)
```

## Stability-to-Risk Ratio

**Business Question:** Evaluate **stability-to-risk ratio** to derive advanced workforce risk intelligence, identify resilient talent segments, and support strategic workforce planning.

```python
# Stability-to-Risk Ratio

df["stability_risk_ratio"] = (
    df["job_security_score"] /
    (df["layoff_risk"] * 100 + 1)
).round(2)
```

## Engineered Metrics Summary

**Business Question:** Evaluate **engineered metrics summary** to derive advanced workforce risk intelligence, identify resilient talent segments, and support strategic workforce planning.

```python
# Engineered Metrics Summary

df[[
    "workforce_stability_index",
    "automation_resilience_score",
    "risk_exposure_index",
    "ai_workforce_readiness_score",
    "stability_risk_ratio"
]].describe().round(2)
```
