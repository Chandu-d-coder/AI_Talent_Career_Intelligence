## Level 1 : Distribution Analytics

Goal: Understand how salary and compensation are distributed across the AI workforce.

This level should contain 5 charts.

```python
import platform
import numpy as np
import pandas as pd
import matplotlib
from matplotlib.ticker import StrMethodFormatter
import matplotlib.pyplot as plt
import seaborn as sns

# Optimize pandas display configurations for comprehensive data inspection.
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", None)

# Log framework metadata and verification metrics.
print("Environment Information")
print("-" * 22)
print(f"Python Version     : {platform.python_version()}")
print(f"Pandas Version     : {pd.__version__}")
print(f"NumPy Version      : {np.__version__}")
print(f"Matplotlib Version : {matplotlib.__version__}")
print(f"Seaborn Version    : {sns.__version__}")
```

```python
# Load target dataset

DATA_PATH = "../../data/processed/cleaned_ai_jobs.csv"
df = pd.read_csv(DATA_PATH)
print("\nDataset Loaded Successfully")

# Execute descriptive matrix shape verification.

print("\nDataset Shape")
print("-" * 16)

total_rows, total_columns = df.shape

print(f"Rows    : {total_rows:,}")
print(f"Columns : {total_columns}")
print("-" * 16)

# Display a quick preview of the dataset schema.

display(df.head())
```

```python
# Create the figure and axes.
fig, ax = plt.subplots(figsize=(15, 7), dpi=150)

# Apply a clean visualization theme.
sns.set_theme(
    style="whitegrid",
    context="notebook",
    font="DejaVu Sans"
)

# Plot the salary distribution.
sns.histplot(
    data=df,
    x="salary_usd",
    bins="fd",
    kde=True,
    stat="count",
    color="#4C72B0",
    edgecolor="white",
    linewidth=0.8,
    alpha=0.90,
    line_kws={
        "linewidth": 2.5,
        "color": "#1F4E79"
    },
    ax=ax
)

# Add a reference line for the median salary.
median_salary = df["salary_usd"].median()

ax.axvline(
    median_salary,
    color="#C44E52",
    linestyle="--",
    linewidth=2
)

# Configure the chart title and axis labels.
ax.set_title(
    "Distribution of Annual Salaries Across AI Job Records",
    fontsize=18,
    fontweight="bold",
    pad=20
)

ax.set_xlabel(
    "Annual Salary (USD)",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

ax.set_ylabel(
    "Number of Job Records",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

# Format the x-axis as currency.
ax.xaxis.set_major_formatter(StrMethodFormatter("${x:,.0f}"))

# Format tick labels.
ax.tick_params(
    axis="both",
    labelsize=11
)

# Improve grid styling.
ax.grid(
    axis="y",
    linestyle="--",
    linewidth=0.8,
    alpha=0.35
)

ax.grid(
    axis="x",
    visible=False
)

# Remove unnecessary spines.
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Optimize layout.
fig.tight_layout()

# Display the visualization.
plt.show()
```
