# Talent Market Intelligence - Visual Analytics

This notebook contains production-ready visualizations for the Talent Market Intelligence module. Each chart includes a brief business objective followed by the visualization code.


## Chart 1: Job Openings Distribution

**Business Objective:** Visualize job openings distribution to understand talent demand and recruitment trends.

```python
import platform
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

# Optimize pandas display configurations for comprehensive data inspection.
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", None)

# Log framework metadata and verification metrics.
print("Environment Information")
print("-" * 22)
print(f"Python Version     : {platform.python_version()}")
print(f"Pandas Version     : {pd.__version__}")
print(f"NumPy Version      : {np.__version__}")
print(f"Matplotlib Version : {matplotlib.__version__}")
print(f"Seaborn Version    : {sns.__version__}")
```

```python
# Load target dataset

DATA_PATH = "../../data/processed/cleaned_ai_jobs.csv"
df = pd.read_csv(DATA_PATH)
print("\nDataset Loaded Successfully")

# Execute descriptive matrix shape verification.

print("\nDataset Shape")
print("-" * 16)

total_rows, total_columns = df.shape

print(f"Rows    : {total_rows:,}")
print(f"Columns : {total_columns}")
print("-" * 16)

# Display a quick preview of the dataset schema.

display(df.head())
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(
    data=df,
    x="job_openings",
    bins=30,
    kde=True,
    color="royalblue",
    edgecolor="white",
    linewidth=0.5,
    alpha=0.85,
    ax=ax
)

ax.set_title("Distribution of Job Openings", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Job Openings")
ax.set_ylabel("Number of Records")
ax.grid(axis="y", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

## Chart 4: Top 10 Countries by Job Openings

**Business Objective:** Visualize top 10 countries by job openings to understand talent demand and recruitment trends.

```python
country_openings = (
    df.groupby("country")["job_openings"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=country_openings,
    x="job_openings",
    y="country",
    color="cornflowerblue",
    ax=ax
)

if ax.containers:
    ax.bar_label(ax.containers[0], fmt="%.1f", padding=3, fontsize=9)

ax.set_title("Top 10 Countries by Average Job Openings", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Average Job Openings")
ax.set_ylabel("Country")
ax.grid(axis="x", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

## Chart 8: Job Openings vs Skill Demand

**Business Objective:** Visualize job openings vs skill demand to understand talent demand and recruitment trends.

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="skill_demand_score",
    y="job_openings",
    alpha=0.35,
    s=35,
    ax=ax
)

ax.set_title("Job Openings vs Skill Demand", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Skill Demand Score")
ax.set_ylabel("Job Openings")
ax.grid(True, linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

```python
# Create Engineered Talent Market Features

df["talent_demand_index"] = (
    (
        df["job_openings"] +
        df["skill_demand_score"] +
        df["ai_adoption_score"]
    ) / 3
).round(2)

df["hiring_efficiency_score"] = (
    df["job_openings"] /
    (df["interview_rounds"] + 1)
).round(2)

df["market_opportunity_index"] = (
    (
        df["job_openings"] +
        df["career_growth_score"] +
        df["salary_percentile"]
    ) / 3
).round(2)

df["skill_premium_index"] = (
    (
        df["skill_demand_score"] *
        df["salary_percentile"]
    ) / 100
).round(2)

df["recruitment_competitiveness_score"] = (
    (
        df["skill_demand_score"] +
        df["hiring_difficulty_score"] +
        (df["interview_rounds"] * 10)
    ) / 3
).round(2)
```

## Chart 9: Talent Demand Index Distribution

**Business Objective:** Visualize talent demand index distribution to understand talent demand and recruitment trends.

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(
    data=df,
    x="talent_demand_index",
    bins=30,
    kde=True,
    color="purple",
    edgecolor="white",
    linewidth=0.5,
    alpha=0.85,
    ax=ax
)

ax.set_title("Distribution of Talent Demand Index", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Talent Demand Index")
ax.set_ylabel("Number of Records")
ax.grid(axis="y", linestyle="--", alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

## Chart 10: Talent Market Correlation Matrix

**Business Objective:** Visualize talent market correlation matrix to understand talent demand and recruitment trends.

```python
# Chart 10: Talent Market Correlation Matrix

talent_corr = (
    df[
        [
            "job_openings",
            "skill_demand_score",
            "hiring_difficulty_score",
            "interview_rounds",
            "talent_demand_index",
            "hiring_efficiency_score",
            "market_opportunity_index",
            "skill_premium_index",
            "recruitment_competitiveness_score"
        ]
    ]
    .corr()
)

fig, ax = plt.subplots(figsize=(10, 8))

sns.heatmap(
    talent_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": 0.8},
    ax=ax
)

ax.set_title(
    "Talent Market Correlation Matrix",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.tight_layout()
plt.show()
```
